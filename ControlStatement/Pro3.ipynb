# Python Programs Using Only if–else Ladder Statements
## (No loops, match-case, or functions)

**Note:** Each program uses only `if`, `elif`, and `else` statements.

```python
# Q1. Blood Donation Eligibility
age = int(input("Enter age: "))
weight = float(input("Enter weight: "))
hb = float(input("Enter hb: "))

if age >= 18 and age <= 65:
    if weight > 50:
        if hb > 12.5:
            print("Eligible for blood donation")
        else:
            print("Not eligible for blood donation")
    else:
        print("Not eligible for blood donation")
else:
    print("Not eligible for blood donation")
```

```python
# Q2. Student Grade Evaluator
marks = float(input("Enter marks: "))

if marks >= 90 and marks <= 100:
    grade = "A+"
elif marks >= 80 and marks <= 89:
    grade = "A"
elif marks >= 70 and marks <= 79:
    grade = "B"
elif marks >= 60 and marks <= 69:
    grade = "C"
elif marks >= 50 and marks <= 59:
    grade = "D"
elif marks >= 0 and marks < 50:
    grade = "Fail"
else:
    grade = "Invalid marks"

print("Grade:", grade)
```

```python
# Q3. Electricity Bill Calculator
units = float(input("Enter units: "))

if units <= 100:
    rate = 5
elif units <= 200:
    rate = 7
elif units <= 300:
    rate = 10
else:
    rate = 15

total_bill = units * rate
print("Total Bill: ₹" + str(total_bill))
```

```python
# Q4. Income Tax Calculator
income = float(input("Enter income: "))

if income <= 250000:
    tax_rate = 0
elif income <= 500000:
    tax_rate = 0.05
elif income <= 1000000:
    tax_rate = 0.20
else:
    tax_rate = 0.30

tax = income * tax_rate
print("Tax to be paid: ₹" + str(tax))
```

```python
# Q5. Temperature Condition
temp = float(input("Input: "))

if temp < 0:
    condition = "Freezing Cold"
elif temp >= 0 and temp <= 10:
    condition = "Very Cold"
elif temp >= 11 and temp <= 20:
    condition = "Cold"
elif temp >= 21 and temp <= 30:
    condition = "Warm"
elif temp >= 31 and temp <= 40:
    condition = "Hot"
else:
    condition = "Extreme Heat"

print("Output:", condition)
```

```python
# Q6. Character Classifier
char = input("Enter character: ")

if len(char) == 1:
    if char >= 'A' and char <= 'Z':
        category = "Uppercase Letter"
    elif char >= 'a' and char <= 'z':
        category = "Lowercase Letter"
    elif char >= '0' and char <= '9':
        category = "Digit"
    else:
        category = "Special Character"
else:
    category = "Invalid input"

print("Output:", category)
```

```python
# Q7. University Admission System
percentage = float(input("Enter percentage: "))
exam_score = float(input("Enter exam score: "))

if percentage >= 90 and exam_score >= 90:
    admission = "Elite Program"
elif percentage >= 80 and exam_score >= 70:
    admission = "Standard Program"
elif percentage >= 60 and exam_score >= 50:
    admission = "Basic Program"
else:
    admission = "Not eligible"

print("Admission in", admission)
```

```python
# Q8. Number Category Analyzer
num = float(input("Enter number: "))

if num > 0:
    if num % 2 == 0:
        category = "Positive Even"
    else:
        category = "Positive Odd"
elif num < 0:
    if num % 2 == 0:
        category = "Negative Even"
    else:
        category = "Negative Odd"
else:
    category = "Zero"

print("Output:", category)
```

```python
# Q9. Shopping Discount System
amount = float(input("Enter purchase amount: "))

if amount < 1000:
    discount_percent = 0
elif amount >= 1000 and amount <= 4999:
    discount_percent = 5
elif amount >= 5000 and amount <= 9999:
    discount_percent = 10
elif amount >= 10000 and amount <= 19999:
    discount_percent = 20
else:
    discount_percent = 30

discount = amount * (discount_percent / 100)
final_amount = amount - discount

print("Discount Applied:", str(discount_percent) + "%")
print("Final Amount: ₹" + str(final_amount))
```

```python
# Q10. Triangle Type Checker
angle1 = float(input("Enter three angles: "))
angle2 = float(input())
angle3 = float(input())

total = angle1 + angle2 + angle3

if total != 180:
    result = "Invalid Triangle"
else:
    if angle1 < 90 and angle2 < 90 and angle3 < 90:
        result = "Triangle is Acute"
    else:
        result = "Triangle is Obtuse"

print("Output:", result)
```
