## P1 — Variables and Addition

Python does not use declarations such as `int x = 10`. Python infers the type from the assigned value. The original also used `a + b`, although the variables were named `x` and `y`.

```python
x = 10
y = 20
print(x + y)
```

## P2 — Convert Input to Integer

`input()` returns text (a string). Convert it to an integer before comparing it with `0`.

```python
x = int(input('Enter num: '))
if x > 0:
    print(x)
```

## P3 — Nested if

The second `if` is inside the first `if`. When `x > 0`, both conditions are true and the number is printed twice. The inner condition is redundant here, but this is a nested `if` example.

```python
x = int(input('Enter a number: '))
if x > 0:
    print(x)
    if x > 0:
        print(x)
```

## P4 — if-else Structure

`else` must be aligned with its matching `if`. With correct indentation, the program prints the value for either case.

```python
x = int(input('Enter a number: '))
if x > 0:
    print(x)
else:
    print(x)
```

## P5 — else Cannot Have a Condition

Use `elif` when a second condition is needed. Use plain `else:` for every remaining case. This version prints `x` whether it is positive or not.

```python
x = int(input('Enter a number: '))
if x > 0:
    print(x)
else:
    print(x)
```

## P6 — Redundant if-else

Both branches print the same value, so the `if-else` is logically unnecessary. It is still valid Python when indented correctly.

```python
x = int(input('Enter a number: '))
if x > 0:
    print(x)
else:
    print(x)
```

## P7 — Separate if Statements

These are three independent conditions. Exactly one condition is true for any integer, so only one `print(x)` runs.

```python
x = int(input('Enter a number: '))
if x > 0:
    print(x)
if x == 0:
    print(x)
if x < 0:
    print(x)
```

## P8 — if-elif-else

This structure tests conditions in sequence and executes only one branch. `x` must be assigned before it is used.

```python
x = int(input('Enter a number: '))
if x > 0:
    print(x)
elif x == 0:
    print(x)
else:
    print(x)
```

## P9 — Indentation in else

The statement inside `else:` must be indented. This corrected program uses proper indentation and assigns an input value to `x`.

```python
x = int(input('Enter a number: '))
if x > 0:
    print(x)
elif x == 0:
    print(x)
else:
    print(x)
```

## P10 — Sum, Subtraction, and if-else

The program calculates the sum and difference. Since `num1` is 10 and `num2` is 20, `num1 > num2` is false, so it prints `subData`, which is 10.

```python
num1 = 10
num2 = 20
sumData = num1 + num2
subData = num2 - num1

if num1 > num2:
    print(sumData)
else:
    print(subData)
```
