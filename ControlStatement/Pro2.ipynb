## Program 1: Maximum between two numbers

```python
def max_of_two():
    a = int(input("Enter first number: "))
    b = int(input("Enter second number: "))
    maximum = a if a > b else b
    print(f"{maximum} is the number between {a} & {b}")

# Example:
# Input: 1 2
# Output: 2 is the number between 1 & 2
max_of_two()
```

## Program 2: Negative, Positive or Zero

```python
def check_sign():
    n = int(input("Enter a number: "))
    if n < 0:
        print(f"{n} is the negative number")
    elif n > 0:
        print(f"{n} is the positive number")
    else:
        print(f"{n} is equal to zero")

# Example:
# Input: -2
# Output: -2 is the negative number
check_sign()
```

## Program 3: Even or Odd

```python
def even_or_odd():
    n = int(input("Enter a number: "))
    if n % 2 == 0:
        print(f"{n} is an Even Number!")
    else:
        print(f"{n} is an Odd Number!")

# Example:
# Input: 4
# Output: 4 is an Even Number!
even_or_odd()
```

## Program 4: Divisible by 5 or not

```python
def divisible_by_5():
    n = int(input("Enter a number: "))
    if n % 5 == 0:
        print(f"{n} is divisible by 5")
    else:
        print(f"{n} is not divisible by 5")

# Example:
# Input: 55
# Output: 55 is divisible by 5
divisible_by_5()
```

## Program 5: Integer (0–6) to weekday (Monday start)

```python
def weekday_from_int():
    days = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]
    n = int(input("Enter an integer (0–6): "))
    if 0 <= n <= 6:
        print(days[n])
    else:
        print("Invalid input. Enter a number between 0 and 6.")

# Example:
# Input: 2
# Output: Wednesday
weekday_from_int()
```

## Program 6: Check if character is an alphabet

```python
def is_alphabet():
    ch = input("Enter a character: ")
    if ch.isalpha() and len(ch) == 1:
        print(f"{ch} is an alphabet.")
    else:
        print(f"{ch} is not an alphabet.")

# Example:
# Input: v
# Output: v is an alphabet.
is_alphabet()
```

## Program 7: Number of days in a month (1–12)

```python
def days_in_month():
    month_days = {
        1: ("January", 31),
        2: ("February", 28),  # Non-leap year default
        3: ("March", 31),
        4: ("April", 30),
        5: ("May", 31),
        6: ("June", 30),
        7: ("July", 31),
        8: ("August", 31),
        9: ("September", 30),
        10: ("October", 31),
        11: ("November", 30),
        12: ("December", 31),
    }
    m = int(input("Enter month number (1–12): "))
    if m in month_days:
        name, days = month_days[m]
        print(f"{name} is a {days}-day month.")
    else:
        print("Invalid month number. Enter 1–12.")

# Example:
# Input: 4
# Output: April is a 30-day month.
days_in_month()
```

## Program 8: Check if number is greater than 10

```python
def greater_than_10():
    n = int(input("Enter a number: "))
    if n > 10:
        print(f"yes {n} is greater than 10")
    else:
        print(f"no {n} is less than 10")

# Example:
# Input: 12
# Output: yes 12 is greater than 10
# Input: 2
# Output: no 2 is less than 10
greater_than_10()
```

## Program 9: Vowel or Consonant

```python
def vowel_or_consonant():
    ch = input("Enter a character: ").lower()
    if ch in "aeiou" and ch.isalpha() and len(ch) == 1:
        print("vowel")
    elif ch.isalpha() and len(ch) == 1:
        print("consonant")
    else:
        print("not a valid alphabet character")

# Example:
# Input: 'a'
# Output: vowel
# Input: 'b'
# Output: consonant
vowel_or_consonant()
```

## Program 10: Leap year check

```python
def is_leap_year():
    year = int(input("Enter year: "))
    # Leap year rule:
    # divisible by 4 AND (not divisible by 100 OR divisible by 400)
    if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
        print(f"{year} is a leap year")
    else:
        print(f"{year} is not a leap year")

# Example:
# Input: 2008
# Output: 2008 is a leap year
# Input: 2013
# Output: 2013 is not a leap year
is_leap_year()
```
