## Q1: Table of a Number


````python
def print_table(n):
    """Prints the multiplication table of a number from 1 to 10."""
    for i in range(1, 11):
        print(f"{n} x {i} = {n*i}")

# Take input from user
num = int(input("Enter a number: "))
print_table(num)
````

## Q2: Even or Odd


````python
def check_even_odd(n):
    """Checks if a number is even or odd."""
    if n % 2 == 0:
        return "Even"
    else:
        return "Odd"

# Take input from user
num = int(input("Enter a number: "))
result = check_even_odd(num)
print(f"{num} is {result}")
````

## Q3: Voting Eligibility


````python
def check_voting_eligibility(age):
    """Checks if a person is eligible for voting (age >= 18)."""
    if age >= 18:
        return "Eligible for voting"
    else:
        return "Not eligible for voting"

# Take input from user
age = int(input("Enter your age: "))
result = check_voting_eligibility(age)
print(f"Age {age}: {result}")
````

## Q4: Arithmetic Operations


````python
def arithmetic_operations(a, b):
    """Prints all arithmetic operations between two numbers."""
    print(f"{a} + {b} = {a+b}")
    print(f"{a} - {b} = {a-b}")
    print(f"{a} * {b} = {a*b}")
    print(f"{a} / {b} = {a/b}")
    print(f"{a} % {b} = {a%b}")
    print(f"{a} // {b} = {a//b}")

# Take input from user
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))
arithmetic_operations(num1, num2)
````

## Q5: Vowel or Consonant


````python
def check_vowel_consonant(ch):
    """Checks if a character is a vowel or consonant."""
    ch = ch.lower()
    if ch in 'aeiou':
        return "Vowel"
    elif ch.isalpha():
        return "Consonant"
    else:
        return "Not a valid alphabet character"

# Take input from user
char = input("Enter a character: ")
result = check_vowel_consonant(char)
print(f"'{char}' is a {result}")
````

## Q6: ASCII Value of Character


````python
def print_ascii(ch):
    """Returns the ASCII value of a character."""
    return ord(ch)

# Take input from user
char = input("Enter a character: ")
ascii_val = print_ascii(char)
print(f"ASCII value of '{char}' is {ascii_val}")
````

## Q7: Maximum of Two Numbers


````python
def max_of_two(a, b):
    """Returns the maximum of two numbers."""
    return a if a > b else b

# Take input from user
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))
maximum = max_of_two(num1, num2)
print(f"Maximum of {num1} and {num2} is {maximum}")
````

## Q9: Average with Variable-Length Arguments (*args)


````python
def average(*args):
    """Calculates the average of any number of values."""
    if len(args) == 0:
        return 0
    return sum(args) / len(args)

# Example usage
print("Average of 2, 4, 6, 8:", average(2, 4, 6, 8))
print("Average of 10, 20, 30:", average(10, 20, 30))
````

### Stack Frame Diagram

For the code:
```python
def outer():
    return "Hello, I'm the inner function!"

ans = outer()
print(ans)
```

**Output:**
```
Hello, I'm the inner function!
```

**Stack Frame Diagram (Text Representation):**

```
Call Stack (Top to Bottom):

|---------------------------|
| outer()                   |
|---------------------------|
| Local variables: (none)   |
| Return value: "Hello..."  |
|---------------------------|
|                           |
|---------------------------|
| <module> (main program)   |
|---------------------------|
| ans = "Hello..."          |
|---------------------------|
```

**Explanation:**
1. The program starts in the `<module>` frame (global scope).
2. When `outer()` is called, a new stack frame for `outer` is pushed on top.
3. `outer()` executes and returns the string.
4. The `outer` frame is popped off the stack.
5. Control returns to `<module>`, where `ans` receives the returned value.
6. `print(ans)` displays the string.


## Q10: Area of Circle


````python
import math

def area_of_circle(r):
    """Returns the area of a circle given its radius."""
    area = math.pi * r * r
    return area

# Take input from user
radius = float(input("Enter radius of the circle: "))
area = area_of_circle(radius)
print(f"Radius: {radius}")
print(f"Area: {area}")
````
