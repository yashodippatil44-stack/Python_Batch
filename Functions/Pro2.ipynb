### Q1. Maximum of 3 numbers

````python
def max_of_three(a, b, c):
    """Return the maximum among 3 numbers."""
    max_num = a
    if b > max_num:
        max_num = b
    if c > max_num:
        max_num = c
    return max_num

# Example
print(max_of_three(3, 7, 5))  # 7
````

### Q2. Square of a number

````python
def square(n):
    """Return the square of a number."""
    return n * n

# Example
print(square(4))  # 16
````

### Q3. Cube of a number

````python
def cube(n):
    """Return the cube of a number."""
    return n * n * n

# Example
print(cube(3))  # 27
````

### Q4. Sum of numbers in a range (while loop)

````python
def sum_in_range(start, end):
    """Return sum of numbers from start to end (inclusive) using a while loop."""
    total = 0
    i = start
    while i <= end:
        total += i
        i += 1
    return total

# Example
print(sum_in_range(1, 5))  # 1+2+3+4+5 = 15
````

### Q5. Average of marks of 5 subjects (while loop)

````python
def average_of_5_marks(marks):
    """
    marks: list/tuple of 5 subject marks.
    Return average using a while loop.
    """
    total = 0
    i = 0
    while i < 5:
        total += marks[i]
        i += 1
    return total / 5

# Example
marks = [80, 90, 70, 85, 95]
print(average_of_5_marks(marks))  # 84.0
````

### Q6. Product of numbers from 1 to n (while loop)

````python
def product_1_to_n(n):
    """Return product of numbers from 1 to n using a while loop."""
    prod = 1
    i = 1
    while i <= n:
        prod *= i
        i += 1
    return prod

# Example
print(product_1_to_n(4))  # 1*2*3*4 = 24
````

### Q7. Factorial of a number (while loop)

````python
def factorial(n):
    """
    Return factorial of n using a while loop.
    factorial of 0 is defined as 1.
    """
    fact = 1
    i = 1
    while i <= n:
        fact *= i
        i += 1
    return fact

# Examples
print(f"factorial of 0 is {factorial(0)}")  # 1
print(f"factorial of 5 is {factorial(5)}")  # 120
````

### Q8. Check prime number (while loop)

````python
def is_prime(n):
    """
    Return True if n is a prime number, else False.
    Uses a while loop.
    """
    if n <= 1:
        return False
    i = 2
    while i * i <= n:
        if n % i == 0:
            return False
        i += 1
    return True

# Examples
print(is_prime(2))   # True
print(is_prime(17))  # True
print(is_prime(18))  # False
````

### Q9. Check composite number (while loop)

````python
def is_composite(n):
    """
    Return True if n is a composite number, else False.
    Composite: integer > 1 that is not prime.
    Uses a while loop.
    """
    if n <= 1:
        return False
    i = 2
    while i * i <= n:
        if n % i == 0:
            return True
        i += 1
    return False

# Examples
print(is_composite(4))   # True
print(is_composite(9))   # True
print(is_composite(13))  # False
````

### Q10. Check perfect number (while loop)

````python
def is_perfect(n):
    """
    Return True if n is a perfect number, else False.
    Perfect number: sum of its proper divisors equals the number.
    Uses a while loop.
    """
    if n <= 1:
        return False
    sum_div = 0
    i = 1
    while i <= n // 2:
        if n % i == 0:
            sum_div += i
        i += 1
    return sum_div == n

# Examples
print(is_perfect(6))    # True (1+2+3 = 6)
print(is_perfect(28))   # True
print(is_perfect(10))   # False
````

## Stack Frame Diagram (for function calls)

When a function is called, the program creates a **stack frame** (also called an activation record) on the call stack. Each stack frame typically contains:

- **Return address**: where to continue after the function finishes
- **Parameters**: the arguments passed to the function
- **Local variables**: variables declared inside the function
- **Temporary values**: intermediate results used during execution

Below is an example stack frame diagram for calling `factorial(5)` which internally uses a while loop.

### Example: `factorial(5)` call

Assume `main()` calls `factorial(5)`:

```
Before calling factorial(5):

|------------------|
| main() frame     |
| - local vars     |
| - return addr    |
|------------------|
      ^
      | stack grows downward (conceptually)

After calling factorial(5) (inside factorial, before returning):

|------------------|
| factorial(5)     |
| - n = 5          |
| - fact = 1       |
| - i = 1          |
| - return addr -> back to main()
|------------------|
| main() frame     |
| - local vars     |
| - return addr    |
|------------------|

During the while loop, `fact` and `i` are updated in the same stack frame:

Iteration 1: i=1, fact=1
Iteration 2: i=2, fact=1*2 = 2
Iteration 3: i=3, fact=2*3 = 6
Iteration 4: i=4, fact=6*4 = 24
Iteration 5: i=5, fact=24*5 = 120
Loop ends (i becomes 6, condition i <= n fails)

When factorial returns 120, its stack frame is popped:

|------------------|
| main() frame     |
| - receives 120   |
|------------------|
```

You can draw this as boxes stacked vertically in your notebook or on paper, labeling each box with:
- Function name
- Parameters
- Local variables
- Return address
