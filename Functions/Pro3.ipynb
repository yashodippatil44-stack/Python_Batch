# Python Assignment using Lambda Functions
## All 10 questions solved with lambda functions and user input

```python
# Q1: Area of square
area_square = lambda side: side * side
side = float(input("Enter side length of square: "))
print(f"Area of square: {area_square(side)}")
```

```python
# Q2: Cube of a number
cube = lambda x: x ** 3
num = float(input("Enter a number: "))
print(f"Cube of {num}: {cube(num)}")
```

```python
# Q3: Maximum of two numbers
max_two = lambda a, b: a if a > b else b
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))
print(f"Maximum of {num1} and {num2}: {max_two(num1, num2)}")
```

```python
# Q4: Area of rectangle
area_rectangle = lambda length, width: length * width
length = float(input("Enter length of rectangle: "))
width = float(input("Enter width of rectangle: "))
print(f"Area of rectangle: {area_rectangle(length, width)}")
```

```python
# Q5: Celsius to Fahrenheit
c_to_f = lambda c: (c * 9/5) + 32
celsius = float(input("Enter temperature in Celsius: "))
print(f"{celsius}C = {c_to_f(celsius)}F")
```

```python
# Q6: Fahrenheit to Celsius
f_to_c = lambda f: (f - 32) * 5/9
fahrenheit = float(input("Enter temperature in Fahrenheit: "))
print(f"{fahrenheit}F = {f_to_c(fahrenheit)}C")
```

```python
# Q7: Last digit of a number
last_digit = lambda n: abs(int(n)) % 10
number = float(input("Enter a number: "))
print(f"Last digit of {number}: {last_digit(number)}")
```

```python
# Q8: Perimeter of square
perimeter_square = lambda side: 4 * side
side = float(input("Enter side length of square: "))
print(f"Perimeter of square: {perimeter_square(side)}")
```

```python
# Q9: Check if string contains 'a'
contains_a = lambda s: 'a' in s.lower()
text = input("Enter a string: ")
if contains_a(text):
    print("The string contains 'a'.")
else:
    print("The string does not contain 'a'.")
```

```python
# Q10: Check if year is leap year
is_leap = lambda year: (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0)
year = int(input("Enter a year: "))
if is_leap(year):
    print(f"{year} is a Leap Year.")
else:
    print(f"{year} is not a Leap Year.")
```
