## Question 1: Arithmetic Operations

```python
x = 15
y = 4

print("Addition:", x + y)
print("Subtraction:", x - y)
print("Multiplication:", x * y)
print("Division:", x / y)
print("Modulus:", x % y)
print("Floor Division:", x // y)
print("Exponentiation:", x ** y)
```

## Question 2: Compound Assignment Operators

```python
num = 50
print("Initial value:", num)

num += 10
print("After += 10:", num)

num -= 5
print("After -= 5:", num)

num *= 2
print("After *= 2:", num)

num /= 4
print("After /= 4:", num)

num %= 5
print("After %= 5:", num)

num //= 3
print("After //= 3:", num)
```

## Question 3: Relational Operators

```python
a = 25
b = 30

print("a = 25 , b = 30")
print("a < b:", a < b)
print("a > b:", a > b)
print("a <= b:", a <= b)
print("a >= b:", a >= b)
print("a == b:", a == b)
print("a != b:", a != b)
```

## Question 4: Logical Operators

```python
x = 10
y = 20
z = 30

print("x = 10 , y = 20 , z = 30")
print(x < y and y < z)
print(x > y or y < z)
print(not (x == y))
print((x < y and y < z) or (x > y or y < z) or not (x == y))
```

## Question 5: Identity Operators (`is`, `is not`) and `id()`

```python
a = 100
b = 100
list1 = [1, 2, 3]
list2 = [1, 2, 3]

print("Integer Variables:")
print("a = 100 , b = 100")
print("id(a) =", id(a), "#(may vary)")
print("id(b) =", id(b), "#(may vary)")
print("a is b:", a is b)
print("a is not b:", a is not b)

print("\nList Variables:")
print("list1 = [1, 2, 3] , list2 = [1, 2, 3]")
print("id(list1) =", id(list1), "#(may vary)")
print("id(list2) =", id(list2), "#(may vary)")
print("list1 is list2:", list1 is list2)
print("list1 == list2:", list1 == list2)
print("list1 is not list2:", list1 is not list2)
```

## Question 6: Membership Operators (`in`, `not in`)

```python
colors = ["red", "blue", "green", "yellow"]

print("colors =", colors)
print('"blue" in colors:', "blue" in colors)
print('"purple" in colors:', "purple" in colors)
print('"pink" not in colors:', "pink" not in colors)
print('"red" not in colors:', "red" not in colors)
```

## Question 7: Bitwise Operations

```python
x = 12
y = 10

print("x = 12 (Binary: 1100)")
print("y = 10 (Binary: 1010)")
print("x & y (Bitwise AND):", x & y)
print("x | y (Bitwise OR):", x | y)
print("x ^ y (Bitwise XOR):", x ^ y)
print("~x (Bitwise NOT):", ~x)
print("x << 2 (Left shift by 2):", x << 2)
print("x >> 2 (Right shift by 2):", x >> 2)
```

## Question 8: Data Types and `type()`

```python
intVal = 10
floatVal = 5.5
stringVal = "Python"

print("Original values:")
print("intVal =", intVal, "Type:", type(intVal))
print("floatVal =", floatVal, "Type:", type(floatVal))
print("stringVal =", stringVal, "Type:", type(stringVal))

result1 = intVal + floatVal
result2 = intVal * floatVal
result3 = floatVal / intVal
result4 = stringVal * 2

print("intVal + floatVal =", result1, "Type:", type(result1))
print("intVal * floatVal =", result2, "Type:", type(result2))
print("floatVal / intVal =", result3, "Type:", type(result3))
print("stringVal * 2 =", result4, "Type:", type(result4))
```

## Question 9: String Concatenation and Repetition

```python
first = "Hello"
second = "World"

print("first =", first)
print("second =", second)
print("Concatenation (first + ' ' + second):", first + " " + second)
print("Repetition (first * 3):", first * 3)
print("Repetition (second * 2):", second * 2)
print("Combined ((first + ' ') * 2 + second):", (first + " ") * 2 + second)
```
