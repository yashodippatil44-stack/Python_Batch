## Program 17: Check Key Existence and Sum Values

```python
# Program 17: Check if key exists and perform operations
data = {0: 10, 1: 20, 2: 25, 3: 30, 4: 40}

# Check if key 2 exists
key_to_check = 2
if key_to_check in data:
    print(f"Key {key_to_check} exists with value: {data[key_to_check]}")

# Sum of values for keys 2 and 3
sum_values = data[2] + data[3]
print(f"Sum of values for keys 2 and 3: {sum_values}")
```

## Program 20: All Data Types

```python
# Program 20: Assign values to all data types

# Integer
int_var = 10
print(f"Integer: {int_var}, Type: {type(int_var)}")

# Float
float_var = 3.14
print(f"Float: {float_var}, Type: {type(float_var)}")

# Complex
complex_var = 2 + 3j
print(f"Complex: {complex_var}, Type: {type(complex_var)}")

# String
str_var = "Hello Python"
print(f"String: {str_var}, Type: {type(str_var)}")

# List
list_var = [1, 2, 3, 4, 5]
print(f"List: {list_var}, Type: {type(list_var)}")

# Tuple
tuple_var = (10, 20, 30)
print(f"Tuple: {tuple_var}, Type: {type(tuple_var)}")

# Dictionary
dict_var = {"name": "Harshal", "age": 21}
print(f"Dictionary: {dict_var}, Type: {type(dict_var)}")

# Set
set_var = {1, 2, 3, 4, 5}
print(f"Set: {set_var}, Type: {type(set_var)}")

# Boolean
bool_var = True
print(f"Boolean: {bool_var}, Type: {type(bool_var)}")

# None
none_var = None
print(f"None: {none_var}, Type: {type(none_var)}")
```
